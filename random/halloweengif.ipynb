```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Polygon, Rectangle, Ellipse
from PIL import Image
import random
import os
from IPython.display import Image as IPImage, HTML

os.makedirs('frames', exist_ok=True)
random.seed(42)
np.random.seed(42)
print("Setup complete 🎃")
```

```python
def draw_haunted_house(ax):
    """Draw the haunted mansion on the hill."""
    # House body
    ax.add_patch(Rectangle((5.8, 2.3), 2.4, 2.2, color='#1a0f0a', zorder=5))
    # Roof (triangle)
    ax.add_patch(Polygon([[5.5, 4.5], [7.0, 5.6], [8.5, 4.5]],
                         closed=True, color='#0d0705', zorder=6))
    # Tower
    ax.add_patch(Rectangle((6.6, 4.5), 0.8, 1.0, color='#120806', zorder=5))
    ax.add_patch(Polygon([[6.5, 5.5], [7.0, 6.3], [7.5, 5.5]],
                         closed=True, color='#0d0705', zorder=6))
    # Glowing windows
    for (wx, wy) in [(6.1, 3.6), (6.1, 2.8), (7.6, 3.6), (7.6, 2.8), (6.9, 4.8)]:
        ax.add_patch(Rectangle((wx, wy), 0.25, 0.35,
                     color='#ffaa33', alpha=0.9, zorder=7))
        # window glow
        ax.add_patch(Circle((wx + 0.125, wy + 0.175), 0.35,
                     color='#ffaa33', alpha=0.15, zorder=4))


def draw_tree(ax, x, scale=1.0):
    """A bare, spooky tree."""
    # Trunk
    ax.add_patch(Rectangle((x - 0.08 * scale, 1.5), 0.16 * scale, 2.0 * scale,
                           color='#0a0605', zorder=4))
    # Branches
    branches = [
        [(x, 3.0), (x - 0.6, 3.6), (x - 0.9, 3.5)],
        [(x, 3.2), (x + 0.5, 3.8), (x + 0.9, 3.7)],
        [(x, 2.7), (x - 0.5, 3.1), (x - 0.8, 3.0)],
        [(x, 2.9), (x + 0.4, 3.3), (x + 0.7, 3.2)],
    ]
    for b in branches:
        xs = [p[0] for p in b]
        ys = [p[1] for p in b]
        ax.plot(xs, ys, color='#0a0605', lw=2.0 * scale, zorder=4)


def draw_fence(ax):
    """Rustic fence across the front."""
    for x in np.arange(0.5, 10, 0.6):
        ax.add_patch(Rectangle((x, 1.6), 0.08, 0.55,
                               color='#050302', zorder=6))
    ax.plot([0.4, 10], [1.75, 1.75], color='#050302', lw=2, zorder=6)
    ax.plot([0.4, 10], [2.05, 2.05], color='#050302', lw=2, zorder=6)


def draw_pumpkin(ax, x, y, glow_intensity=0.5, scale=1.0):
    """A glowing jack-o'-lantern."""
    # Body (three overlapping ellipses for a pumpkin shape)
    ax.add_patch(Ellipse((x, y), 0.9 * scale, 0.7 * scale,
                         color='#c9521a', zorder=8))
    ax.add_patch(Ellipse((x - 0.22 * scale, y), 0.5 * scale, 0.68 * scale,
                         color='#e0631f', zorder=8))
    ax.add_patch(Ellipse((x + 0.22 * scale, y), 0.5 * scale, 0.68 * scale,
                         color='#e0631f', zorder=8))
    # Stem
    ax.add_patch(Rectangle((x - 0.05 * scale, y + 0.32 * scale),
                           0.1 * scale, 0.12 * scale,
                           color='#2b1a08', zorder=9))
    # Glowing outer halo
    ax.add_patch(Circle((x, y), 0.7 * scale, color='#ff8822',
                        alpha=0.12 * glow_intensity, zorder=3))
    ax.add_patch(Circle((x, y), 1.1 * scale, color='#ff6600',
                        alpha=0.05 * glow_intensity, zorder=3))
    # Face — eyes (triangles)
    eye_color = '#ffcc33'
    ax.add_patch(Polygon([[x - 0.22 * scale, y + 0.18 * scale],
                          [x - 0.05 * scale, y + 0.18 * scale],
                          [x - 0.14 * scale, y + 0.02 * scale]],
                         closed=True, color=eye_color, zorder=10))
    ax.add_patch(Polygon([[x + 0.05 * scale, y + 0.18 * scale],
                          [x + 0.22 * scale, y + 0.18 * scale],
                          [x + 0.14 * scale, y + 0.02 * scale]],
                         closed=True, color=eye_color, zorder=10))
    # Mouth (jagged smile)
    ax.add_patch(Polygon([
        [x - 0.25 * scale, y - 0.05 * scale],
        [x - 0.15 * scale, y - 0.18 * scale],
        [x - 0.05 * scale, y - 0.05 * scale],
        [x + 0.05 * scale, y - 0.18 * scale],
        [x + 0.15 * scale, y - 0.05 * scale],
        [x + 0.25 * scale, y - 0.15 * scale],
        [x + 0.10 * scale, y - 0.28 * scale],
        [x - 0.10 * scale, y - 0.28 * scale],
    ], closed=True, color=eye_color, zorder=10))


def draw_bat(ax, x, y, scale=1.0, flap=0.0):
    """A flying bat with flapping wings."""
    wing_y = 0.18 * flap * scale
    # Body
    ax.add_patch(Ellipse((x, y), 0.12 * scale, 0.16 * scale,
                         color='#050505', zorder=12))
    # Left wing
    ax.add_patch(Polygon([
        [x - 0.05 * scale, y],
        [x - 0.35 * scale, y + wing_y + 0.06 * scale],
        [x - 0.42 * scale, y + wing_y - 0.02 * scale],
        [x - 0.20 * scale, y - 0.05 * scale],
    ], closed=True, color='#050505', zorder=12))
    # Right wing
    ax.add_patch(Polygon([
        [x + 0.05 * scale, y],
        [x + 0.35 * scale, y + wing_y + 0.06 * scale],
        [x + 0.42 * scale, y + wing_y - 0.02 * scale],
        [x + 0.20 * scale, y - 0.05 * scale],
    ], closed=True, color='#050505', zorder=12))


def draw_moon(ax, x, y, radius=1.1):
    """A glowing full moon."""
    # Halo layers
    ax.add_patch(Circle((x, y), radius * 2.4, color='#fff5cc',
                        alpha=0.05, zorder=2))
    ax.add_patch(Circle((x, y), radius * 1.6, color='#fff2b3',
                        alpha=0.10, zorder=2))
    ax.add_patch(Circle((x, y), radius * 1.15, color='#fff8d9',
                        alpha=0.22, zorder=2))
    # Moon itself
    ax.add_patch(Circle((x, y), radius, color='#f5e9c0', zorder=3))
    # Craters
    for (cx, cy, cr) in [(-0.3, 0.2, 0.15), (0.25, -0.15, 0.12), (0.0, 0.35, 0.08)]:
        ax.add_patch(Circle((x + cx * radius, y + cy * radius),
                            cr * radius, color='#e0d4a8', alpha=0.6, zorder=3))


def draw_leaves(ax, t):
    """Falling leaves scattered across the scene."""
    for i in range(25):
        lx = (i * 1.7 + t * 0.4 + np.sin(t * 0.5 + i) * 0.3) % 10
        ly = (2.5 - (t * 0.15 + i * 0.11) % 2.5)
        # Leaf color (autumn palette)
        colors = ['#c9521a', '#e08a1e', '#8a3a10', '#d4a017']
        ax.add_patch(Ellipse((lx, ly), 0.14, 0.08,
                             angle=(t * 30 + i * 20) % 360,
                             color=colors[i % len(colors)],
                             alpha=0.85, zorder=11))


def draw_sky(ax):
    """Deep gradient night sky."""
    grad = np.linspace(0, 1, 256).reshape(-1, 1)
    grad = np.tile(grad, (1, 10))
    ax.imshow(grad, extent=[0, 10, 0, 6], aspect='auto',
              cmap='twilight_shifted', alpha=0.55, zorder=0)
    # Ground
    ax.add_patch(Rectangle((0, 0), 10, 2.0, color='#0a0604', zorder=1))
    # Foggy hill behind the house
    ax.add_patch(Ellipse((7, 2.0), 7, 1.5, color='#0f0805',
                         alpha=0.85, zorder=2))
```

```python
def render_frame(t, total_frames, filename):
    """Draw one frame at time t (0..1) and save it."""
    fig, ax = plt.subplots(figsize=(8, 6), dpi=100)
    ax.set_xlim(0, 10)
    ax.set_ylim(0, 6)
    ax.axis('off')
    fig.patch.set_facecolor('#02010a')

    # Sky + ground
    draw_sky(ax)

    # Moon drifting slowly across the sky
    moon_x = 1.6 + 1.2 * np.sin(t * 2 * np.pi * 0.3)
    moon_y = 4.6 + 0.15 * np.sin(t * 2 * np.pi * 0.5)
    draw_moon(ax, moon_x, moon_y, radius=1.0)

    # Trees
    draw_tree(ax, 1.0, scale=1.1)
    draw_tree(ax, 3.2, scale=0.85)

    # Haunted house
    draw_haunted_house(ax)

    # Fence
    draw_fence(ax)

    # Falling leaves
    draw_leaves(ax, t * 20)

    # Flying bats — sinusoidal paths across the sky
    bat_positions = [
        (1.0 + 8.0 * ((t * 0.8 + 0.0) % 1.0), 4.6 + 0.4 * np.sin(t * 6)),
        (1.0 + 8.0 * ((t * 0.6 + 0.4) % 1.0), 5.0 + 0.5 * np.sin(t * 5 + 1)),
        (1.0 + 8.0 * ((t * 1.1 + 0.7) % 1.0), 4.2 + 0.3 * np.sin(t * 7 + 2)),
        (1.0 + 8.0 * ((t * 0.5 + 0.2) % 1.0), 5.3 + 0.4 * np.sin(t * 4 + 3)),
    ]
    for i, (bx, by) in enumerate(bat_positions):
        flap = np.sin(t * 2 * np.pi * 4 + i)
        draw_bat(ax, bx, by, scale=0.9, flap=flap)

    # Glowing jack-o'-lantern in the foreground
    glow = 0.7 + 0.3 * np.sin(t * 2 * np.pi * 2)  # pulsing glow
    draw_pumpkin(ax, 2.0, 0.9, glow_intensity=glow, scale=1.3)

    # "Happy Halloween" text
    ax.text(5.0, 5.7, "Happy Halloween", color='#ffaa33',
            fontsize=20, ha='center', va='center',
            fontweight='bold', alpha=0.95, zorder=15,
            family='serif')

    plt.tight_layout(pad=0)
    plt.savefig(filename, facecolor='#02010a', bbox_inches='tight', pad_inches=0)
    plt.close(fig)


# --- Render the frames ---
NUM_FRAMES = 60
for i in range(NUM_FRAMES):
    t = i / NUM_FRAMES
    render_frame(t, NUM_FRAMES, f'frames/frame_{i:03d}.png')
    if i % 10 == 0:
        print(f"Rendered frame {i}/{NUM_FRAMES}")

print("✅ All frames rendered!")
```

```python
frame_files = sorted([f'frames/{f}' for f in os.listdir('frames') if f.endswith('.png')])
images = [Image.open(f).convert('RGB') for f in frame_files]

output_gif = 'halloween.gif'
images[0].save(
    output_gif,
    save_all=True,
    append_images=images[1:],
    duration=80,       # ms per frame (~12.5 fps)
    loop=0,            # infinite loop
    optimize=True,
    disposal=2,
)

print(f"🎃 GIF saved: {output_gif} ({os.path.getsize(output_gif)/1024:.1f} KB)")
```
