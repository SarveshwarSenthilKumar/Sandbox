```python
from PIL import Image, ImageDraw, ImageFont
import numpy as np
import os
from IPython.display import Image as IPImage, HTML
```

```python
# --- Config ---
num_frames = 30
width, height = 400, 400
duration_ms = 100          # ms per frame (100ms = 10 fps)
output_path = 'animation.gif'

frames = []

for i in range(num_frames):
    # Create a blank white frame
    img = Image.new('RGB', (width, height), 'white')
    draw = ImageDraw.Draw(img)

    # Animate a circle moving horizontally across the canvas
    x = int(50 + (width - 100) * i / num_frames)
    y = height // 2
    radius = 40
    draw.ellipse(
        [x - radius, y - radius, x + radius, y + radius],
        fill='cornflowerblue'
    )

    # Add a frame counter in the top-left corner
    draw.text((20, 20), f"Frame {i + 1}/{num_frames}", fill='black')

    frames.append(img)

print(f"Created {len(frames)} frames")
```

```python
frames[0].save(
    output_path,
    save_all=True,
    append_images=frames[1:],
    duration=duration_ms,   # ms per frame
    loop=0,                 # 0 = loop forever
    optimize=False
)

print(f"GIF saved to {output_path} ({os.path.getsize(output_path)} bytes)")
```
